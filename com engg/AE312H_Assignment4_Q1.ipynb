## Q1. Aircraft Pitching-Moment Characteristics (Crout LU Decomposition)

Model: $C_m = C_{m_0} + C_{m_\alpha}\alpha + C_{m_{\delta_e}}\delta_e$ (angles in radians).

### Tasks 1–2: Dynamic pressure and $C_m$

```python
import numpy as np
np.set_printoptions(precision=6, suppress=True)

rho, S, c = 1.225, 0.5, 0.3
V     = np.array([50.0, 60.0, 70.0])
alpha = np.deg2rad([2.0, 6.0, 10.0])
de    = np.deg2rad([-1.0, 2.0, 5.0])
M     = np.array([8.2773, -5.9759, -32.4914])

q  = 0.5*rho*V**2          # Task 1
Cm = M/(q*S*c)             # Task 2

for i in range(3):
    print(f"Test {i+1}: q = {q[i]:10.4f} Pa,  Cm = {Cm[i]: .6f}")
```

### Tasks 3–4: Equations and matrix form $Ax=b$
Each test gives: $C_{m_0} + \alpha_i\,C_{m_\alpha} + \delta_{e,i}\,C_{m_{\delta_e}} = C_{m,i}$

```python
A = np.column_stack([np.ones(3), alpha, de])
b = Cm.copy()

for i in range(3):
    print(f"Eq {i+1}: Cm0 + ({alpha[i]:.6f}) Cmα + ({de[i]:.6f}) Cmδe = {b[i]:.6f}")
print("\nA =\n", A)
print("b =", b)
```

### Task 5: Crout's algorithm ($L$ general lower, $U$ unit upper triangular)

$$l_{ij}=a_{ij}-\sum_{k=1}^{j-1} l_{ik}u_{kj}\ (i\ge j),\qquad u_{ji}=\frac{a_{ji}-\sum_{k=1}^{j-1} l_{jk}u_{ki}}{l_{jj}}\ (i>j),\ u_{jj}=1$$

**Important:** here the pivot $l_{33}$ comes out as $\approx 0$, i.e. $A$ is singular (see the note after the output).

```python
TOL = 1e-10

def crout(A, verbose=True):
    n = A.shape[0]
    L = np.zeros((n, n)); U = np.eye(n)
    for j in range(n):
        for i in range(j, n):                      # column j of L
            s = sum(L[i, k]*U[k, j] for k in range(j))
            L[i, j] = A[i, j] - s
            if verbose:
                print(f"L[{i+1},{j+1}] = A[{i+1},{j+1}] - {s:.6f} = {L[i,j]:.6e}")
        if abs(L[j, j]) < TOL:
            print(f">>> pivot L[{j+1},{j+1}] = {L[j,j]:.2e} ~ 0  ->  A is SINGULAR")
            continue                               # cannot divide; U row stays 'free'
        for i in range(j+1, n):                    # row j of U
            s = sum(L[j, k]*U[k, i] for k in range(j))
            U[j, i] = (A[j, i] - s)/L[j, j]
            if verbose:
                print(f"U[{j+1},{i+1}] = (A[{j+1},{i+1}] - {s:.6f}) / L[{j+1},{j+1}] = {U[j,i]:.6f}")
        if verbose: print("-"*50)
    return L, U

L, U = crout(A)
print("\nL =\n", L)
print("\nU =\n", U)
print("det(A) =", np.linalg.det(A), "  rank(A) =", np.linalg.matrix_rank(A, tol=1e-10))
```

**Why singular?** The test points satisfy $\delta_e = 0.75\,\alpha - 2.5^\circ$ exactly (2→-1, 6→2, 10→5), so the $\delta_e$ column of $A$ equals $0.75\times$(α column) $-2.5^\circ\times$(column of ones). The three tests cannot separate the effects of $\alpha$ and $\delta_e$. The data themselves are still self-consistent (the measured $C_m$ is exactly linear in $\alpha$), so the system has **infinitely many solutions** – a one-parameter family – rather than none.

### Task 6: Forward substitution, $Ly=b$
(A zero pivot is allowed only if the equation is consistent, $b_i - \sum l_{ik}y_k \approx 0$; then $y_i$ is a free parameter.)

```python
def forward_sub(L, b, verbose=True):
    n = len(b); y = np.zeros(n)
    for i in range(n):
        s = sum(L[i, k]*y[k] for k in range(i))
        if abs(L[i, i]) < TOL:
            resid = b[i] - s
            print(f"y[{i+1}]: 0*y = {resid:.2e}  -> consistent (~0), y[{i+1}] is FREE" if abs(resid) < 1e-5
                  else f"y[{i+1}]: 0*y = {resid:.2e}  -> INCONSISTENT, no solution")
            y[i] = np.nan                           # free parameter
        else:
            y[i] = (b[i] - s)/L[i, i]
            if verbose: print(f"y[{i+1}] = ({b[i]:.6f} - {s:.6f}) / {L[i,i]:.6f} = {y[i]:.6f}")
    return y

y = forward_sub(L, b)
```

### Task 7: Backward substitution, $Ux=y$
Because $U$ has a unit diagonal, the free parameter is $x_3 = C_{m_{\delta_e}} = t$. We solve for $C_{m_0}, C_{m_\alpha}$ as functions of $t$.

```python
def backward_sub(U, y, t):
    n = len(y); x = np.zeros(n)
    y = np.where(np.isnan(y), t, y)                 # free parameter
    for i in range(n-1, -1, -1):
        s = sum(U[i, k]*x[k] for k in range(i+1, n))
        x[i] = (y[i] - s)/U[i, i]
    return x

# Family of solutions: x(t) = x(0) + t * d   (d = null-space direction)
x0 = backward_sub(U, y, 0.0)
d  = backward_sub(U, y, 1.0) - x0
print("Cm0   = %.6f + %.6f * t" % (x0[0], d[0]))
print("Cmα   = %.6f + %.6f * t" % (x0[1], d[1]))
print("Cmδe  = t   (free)")
print("null-space check  A@d =", A @ d)

# Pick t = Cmδe = -0.3 (gives the 'clean' values Cm0=0.05, Cmα=-0.55)
t = -0.3
x = backward_sub(U, y, t)
Cm0, Cma, Cmde = x
print(f"\nWith t = {t}:")
print(f"Cm0  = {Cm0:.6f}")
print(f"Cmα  = {Cma:.6f} per rad")
print(f"Cmδe = {Cmde:.6f} per rad")
```

### Task 8: Verification, $A-LU$
Because of the zero pivot, $LU$ reproduces $A$ with $L_{33}=0$ (the factorization is still valid, just not invertible).

```python
R = A - L @ U
print("A - LU =\n", R)
print("||A - LU||_inf =", np.linalg.norm(R, np.inf))
print("Residual of Ax-b :", A @ x - b)
```

### Task 9: Prediction at $\alpha=8^\circ,\ \delta_e=3^\circ$

Because $\delta_e=3^\circ$ is *off* the line $\delta_e=0.75\alpha-2.5^\circ$ (which would give 3.5° at α = 8°), the prediction depends on the free parameter $t$.

```python
a_p, d_p = np.deg2rad(8.0), np.deg2rad(3.0)
Cm_pred = Cm0 + Cma*a_p + Cmde*d_p
print(f"Predicted Cm (t = {t}) = {Cm_pred:.6f}")

print("\nSensitivity to the free parameter:")
for tt in [-0.6, -0.3, 0.0, 0.3]:
    xx = backward_sub(U, y, tt)
    print(f"  Cmδe = {tt:5.1f}  ->  Cm_pred = {xx[0] + xx[1]*a_p + xx[2]*d_p: .6f}")
```
