```python
import numpy as np

rho, S, c = 1.225, 0.5, 0.3
```

```python
#Tasks 1 and 2

#condition 1
V1, alpha1, delta1, M1 = 50, np.deg2rad(2), np.deg2rad(-1), 8.2773
q1 = 0.5 * rho * (V1**2)
Cm1 = M1 / (q1 * S * c)

#condition 2
V2, alpha2, delta2, M2 = 60, np.deg2rad(6), np.deg2rad(2), -5.9759
q2 = 0.5 * rho * (V2**2)
Cm2 = M2 / (q2 * S * c)

#condition 3
V3, alpha3, delta3, M3 = 70, np.deg2rad(10), np.deg2rad(5), -32.4914
q3 = 0.5 * rho * (V3**2)
Cm3 = M3 / (q3 * S * c)

print(f"Condition 1: q = {q1:.2f} Pa, Cm = {Cm1:.4f}")
print(f"Condition 2: q = {q2:.2f} Pa, Cm = {Cm2:.4f}")
print(f"Condition 3: q = {q3:.2f} Pa, Cm = {Cm3:.4f}")
```

```python
#Task 3

print(f"Cm_0 + Cm_alpha * {alpha1} + Cm_delta * {delta1} = {Cm1}")
print(f"Cm_0 + Cm_alpha * {alpha2} + Cm_delta * {delta2} = {Cm2}")
print(f"Cm_0 + Cm_alpha * {alpha3} + Cm_delta * {delta3} = {Cm3}")
```

```python
#Task 4

Cm_0, Cm_alpha, Cm_delta = 0, 0, 0

A = np.array([[1, alpha1, delta1],
              [1, alpha2, delta2],
              [1, alpha3, delta3]])

x = np.array([Cm_0, Cm_alpha, Cm_delta])

b = np.array([Cm1, Cm2, Cm3])

print(f"Matrix A:\n{A}")
print(f"Vector b:\n{b}")
```

```python
#Task 5

n = 3

Ac = np.zeros((n, n))

for j in range(0, n):
    for i in range(0, j+1):
        Ac[i, j] = A[i, j] - np.sum(Ac[i, :i] * Ac[:i, j])
        print(f"U[{i},{j}] = A[{i},{j}] - sum(L[{i},:{i}] * U[:{i},{j}]) = {Ac[i, j]}")
    for i in range(j+1, n):
        Ac[i, j] = (1 / Ac[j, j]) * (A[i, j] - np.sum(Ac[i, :j] * Ac[:j, j]))
        print(f"L[{i},{j}] = (A[{i},{j}] - sum(L[{i},:{j}] * U[:{j},{j}])) / U[{j},{j}] = {Ac[i, j]}")

print(f"Matrix Ac:\n{Ac}")

L = np.zeros((n, n))
U = np.zeros((n, n))

for i in range(n):
    for j in range(n):
        if i > j:
            L[i, j] = Ac[i, j]
        elif i == j:
            L[i, j] = 1
            U[i, j] = Ac[i, j]
        else:
            U[i, j] = Ac[i, j]

print(f"Matrix L:\n{L}")
print(f"Matrix U:\n{U}")
```

```python
#Task 6

y = np.zeros(n)
for i in range(n):
    y[i] = b[i] - np.sum(Ac[i, :i] * y[:i])

print(f"Vector y:\n{y}")
```

```python
#Task 7

x = np.zeros(n)
for i in range(n-1, -1, -1):
    x[i] = (y[i] - np.sum(Ac[i, i+1:] * x[i+1:])) / Ac[i, i]

print(f"Vector x:\n{x}")
```

```python
#Task 8

R = A - L @ U
print(f"Matrix R:\n{R}")
```
