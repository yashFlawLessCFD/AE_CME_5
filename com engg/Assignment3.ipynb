quasi-one-dimensional, isentropic nozzle

R=A/A*

```python
import numpy as np
gamma = 1.4
R = 2.5
tol = 10e-6
M_u = 1
M_l = 0.1
M = M_l
er = 1

def f(M):
    return R - (1/M)*((2/(gamma+1))*(1 + (gamma-1)/2 * M**2))**((gamma+1)/(2*(gamma-1)))

i = 1
while er > tol:

    M_old = M
    M = (M_u + M_l)/2
    if f(M) * f(M_u) < 0:
        M_l = M
    else:
        M_u = M
    er = abs((M - M_old) / M)
    print(f"Error at iteration {i}: {er:.6f}")
    i += 1

print("M =", M)
print("f(M) =", f(M))
```

Using Newton Rapson

```python
gamma = 1.4
R = 2.5
er = 1
tol = 10e-6
M = 0.5       

def f(M):
    return R - (1/M)*((2/(gamma+1))*(1 + (gamma-1)/2 * M**2))**((gamma+1)/(2*(gamma-1)))

def f1(M):
    return ((((2/(gamma+1))*(1 + (gamma-1)/2 * M**2))**(gamma+1)/(2*(gamma-1)))/(M**2)- (1/M) * (gamma+1)/(2*(gamma-1))* ((2/(gamma+1))*(1 + (gamma-1)/2 * M**2))**((gamma+1)/(2*(gamma-1))-1)* (2/(gamma+1))*(gamma-1)*M)


i = 1
while er > tol:
    M_old = M
    M = M_old - f(M_old)/f1(M_old)
    er = abs((M - M_old)/M)
    print(f"Error at iteration {i}: {er:.6e}")
    print(f"M = {M:.6f}")
    i += 1


print("\nFinal result:")
print("M =", M)
print("f(M) =", f(M))
```
