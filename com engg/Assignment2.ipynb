```python
import numpy as np
g = 9.81
c=15
v=36
t=10
e_s=1
m_l=50
m_u=100
m_old=m_l

def f(m):
    return v-(g*m/c)*(1-np.exp(-(c/m)*t))

while e_s>0.001 :
    m=(f(m_u)*m_l-f(m_l)*m_u)/(f(m_u)-f(m_l))
    if f(m)*f(m_u)<0:
        m_l=m
    else :
        m_u=m
    e_s = abs((m-m_old)/m)
    m_old = m
    print(e_s)
    
print(f(m))
print(f(m_u))
print(f(m_l))
print(m)
```

Question 2(a) bisection method

```python
import numpy as np
rho=1.23
mu=1.790e-5
D=0.005
V=40
E=0.0000015
Re= (rho*V*D)/mu
f_l=0.008
f_u=0.08
f_old=f_u
er=5
def F(f):
    return 1/np.sqrt(f)+2*np.log10((E/(3.7*D))+(2.51/(Re*np.sqrt(f))))

i=1 
while er>0.00001:
    f=(f_u+f_l)/2
    F(f)
    if F(f)*F(f_u)<0:
        f_l=f
    else:
        f_u=f
    er=abs((f-f_old)/f)
    f_old=f
    print(f"Error at iteration {i}: {er:.6f}")
    i += 1
print(f"F(f): {F(f)}")
print(f"f: {f}")
```
