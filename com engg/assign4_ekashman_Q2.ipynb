```python
import numpy as np
```

```python
#Task 1

CL = np.array([0.15, 0.35, 0.55, 0.85])
CD = np.array([0.01934625, 0.02464125, 0.03437625, 0.05835375])
n = 4

for i in range(n):
    print(f"CD0 + k1 * {CL[i]} + k2 * {CL[i]**2} + k3 * {CL[i]**3} = {CD[i]}")
```

```python
#Task 2

CD0, k1, k2, k3 = 0, 0, 0, 0

A = np.zeros((n, n))
for i in range(n):
    A[i, :] = [1, CL[i], CL[i]**2, CL[i]**3]

x = np.array([CD0, k1, k2, k3])
b = CD.copy()

print(f"Matrix A:\n{A}")
print(f"Vector b:\n{b}")
```

```python
#Task 3

Ac = np.zeros((n, n))

for j in range(0, n):
    for i in range(0, j+1):
        Ac[i, j] = A[i, j] - np.sum(Ac[i, :i] * Ac[:i, j])
        print(f"U[{i},{j}] = A[{i},{j}] - sum(L[{i},:{i}] * U[:{i},{j}]) = {Ac[i, j]}")
    for i in range(j+1, n):
        Ac[i, j] = (1 / Ac[j, j]) * (A[i, j] - np.sum(Ac[i, :j] * Ac[:j, j]))
        print(f"L[{i},{j}] = (A[{i},{j}] - sum(L[{i},:{j}] * U[:{j},{j}])) / U[{j},{j}] = {Ac[i, j]}")

print(f"Matrix Ac:\n{Ac}")

L = np.zeros((n, n))
U = np.zeros((n, n))

for i in range(n):
    for j in range(n):
        if i > j:
            L[i, j] = Ac[i, j]
        elif i == j:
            L[i, j] = 1
            U[i, j] = Ac[i, j]
        else:
            U[i, j] = Ac[i, j]

print(f"Matrix L:\n{L}")
print(f"Matrix U:\n{U}")
```

```python
#Task 4

y = np.zeros(n)
for i in range(n):
    y[i] = b[i] - np.sum(Ac[i, :i] * y[:i])

print(f"Vector y:\n{y}")
```

```python
#Task 5

x = np.zeros(n)
for i in range(n-1, -1, -1):
    x[i] = (y[i] - np.sum(Ac[i, i+1:] * x[i+1:])) / Ac[i, i]

print(f"Vector x:\n{x}")
```

```python
#Task 6

print(f"CD0 = {x[0]}, k1 = {x[1]}, k2 = {x[2]}, k3 = {x[3]}")
```

```python
#Task 7

R = A - L @ U
norm_R = np.max(np.sum(np.abs(R), axis=1))   # infinity norm = max absolute row sum

print(f"Matrix R:\n{R}")
print(f"||R||_inf = {norm_R}")
```

```python
#Task 8

r = A @ x - b
eps_r = np.max(np.abs(r)) / np.max(np.abs(b))

print(f"Relative residual = {eps_r}")
```
