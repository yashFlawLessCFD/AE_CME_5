Prblm 1: Range of x is from -2 to 2

```python
x = np.linspace(-2, 2, 400) # defining domain of x, here x can be any ral value
y = (x**3-2*x+1)/(x**2+1)
plt.figure(figsize=(8, 5))
plt.plot(x, y, label=r'$f(x) = \frac{x^3 - 2x + 1}{x^2 + 1}$', color='b')
plt.title('In the range of x, F(x)')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define grid and parameters
a = 1.0     # Cylinder radius
u0 = 1.0    # Free-stream velocity

x = np.linspace(-3, 3, 400)
y = np.linspace(-3, 3, 400)
X, Y = np.meshgrid(x, y)
R = np.sqrt(X**2 + Y**2)
THETA = np.arctan2(Y, X)

# Mask inside of the cylinder
mask = R < a

# Velocity potential function phi
Phi = u0 * (R - (a**2 / R)) * np.sin(THETA)
Phi[mask] = np.nan

# Plot constant potential lines (equipoteintials)
plt.figure(figsize=(7, 6))
cs = plt.contour(X, Y, Phi, levels=np.linspace(-4, 4, 20), colors='blue', linestyles='dashed')
plt.clabel(cs, inline=True, fontsize=8)

# Add cylinder boundary
cylinder = plt.Circle((0, 0), a, color='gray', alpha=0.5)
plt.gca().add_patch(cylinder)

plt.xlabel('x')
plt.ylabel('y')
plt.title('Lines of Constant Velocity Potential ($\phi$)')
plt.axis('equal')
plt.grid(True)
plt.show()
```
