## popularity based 

```python
rating_with_name= ratings.merge(books, on='ISBN')
```

```python
num_rating_df = rating_with_name.groupby('Book-Title').count()['Book-Rating'].reset_index()
num_rating_df.rename(columns={'Book-Rating':'num_rating'},inplace=True)
num_rating_df
```

```python
avg_rating_df = rating_with_name.groupby('Book-Title')['Book-Rating'].mean().reset_index()
avg_rating_df.rename(columns={'Book-Rating':'avg_rating'},inplace=True)
avg_rating_df
```

```python
popularity_df = num_rating_df.merge(avg_rating_df, on='Book-Title' )
popularity_df
```

```python
popularity_df= popularity_df[popularity_df['num_rating']>=250].sort_values('avg_rating', ascending=False).head(50)
```

```python
popularity_df.merge(books, on='Book-Title').drop_duplicates('Book-Title')
```

## Collabrative filtering based

```python
x = rating_with_name.groupby('User-ID').count()['Book-Rating']>200
padhe_likhe_gawar= x[x].index ##bool indexing
```

```python
filtered_rating = rating_with_name[rating_with_name['User-ID'].isin(padhe_likhe_gawar)]
```

```python
y=filtered_rating.groupby('Book-Title').count()['Book-Rating']>=50
Famous_Book= y[y].index
```

```python
final_rating = filtered_rating[filtered_rating['Book-Title'].isin(Famous_Book)]
```

```python
pt= final_rating.pivot_table(index='Book-Title', columns='User-ID', values='Book-Rating')
```

```python
pt.fillna(0, inplace=True)
```

```python
pt
```

```python
!pip install scikit-learn
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity_score = cosine_similarity(pt)
```

```python
def recommend(Book_name):
    #index fetch
    matches = np.where(pt.index.str.lower()==Book_name.lower())[0]
    if len(matches) == 0:
        print(f"Book '{Book_name}' not found in dataset. Please check the spelling.")
        return
    index = matches[0]
    similar_items = sorted(list(enumerate(similarity_score[index])),key=lambda x:x[1],reverse=True)[1:6]
    for i in similar_items:
        print(pt.index[i[0]])
```

```python
recommend('Message in a Bottle')
```

```python
recommend('a walk to remember')
```

```python
recommend('the notebook')
```

```python
recommend('Blood Meridian')
```
